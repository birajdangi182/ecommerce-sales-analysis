# E-commerce Sales Analysis

**Goal:** Explore sales, profit, products, customers and regional performance.

The included CSV is a **synthetic practice dataset** created for this portfolio example. It is not company-confidential or real customer data.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("../data/sales_data.csv", parse_dates=["order_date"])
df.head()
```

## 1. Data quality checks

```python
print("Rows and columns:", df.shape)
print("\nMissing values:\n", df.isna().sum())
print("Duplicate order IDs:", df["order_id"].duplicated().sum())
df.describe(include="all")
```

## 2. Business KPIs

```python
total_revenue = df["revenue"].sum()
total_profit = df["profit"].sum()
total_orders = df["order_id"].nunique()
unique_customers = df["customer_id"].nunique()

kpis = pd.Series({
    "Revenue": total_revenue,
    "Profit": total_profit,
    "Orders": total_orders,
    "Unique customers": unique_customers,
    "Profit margin (%)": total_profit / total_revenue * 100,
    "Revenue per order": total_revenue / total_orders
})
kpis
```

## 3. Monthly trend

```python
df["month"] = df["order_date"].dt.to_period("M").astype(str)
monthly = df.groupby("month", as_index=False).agg(
    revenue=("revenue", "sum"),
    profit=("profit", "sum")
)

plt.figure(figsize=(11, 4))
plt.plot(monthly["month"], monthly["revenue"], marker="o")
plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 4. Category and region performance

```python
category = df.groupby("category", as_index=False).agg(
    revenue=("revenue", "sum"),
    profit=("profit", "sum")
).sort_values("revenue", ascending=False)

display(category)

plt.figure(figsize=(8, 4))
plt.bar(category["category"], category["revenue"])
plt.title("Revenue by Category")
plt.ylabel("Revenue")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

## 5. Findings (recalculate after running)

```python
best_month = monthly.loc[monthly["revenue"].idxmax()]
top_category = category.loc[category["revenue"].idxmax(), "category"]
print(f"Highest-revenue month: {best_month['month']}")
print(f"Highest-revenue category: {top_category}")
print(f"Overall profit margin: {total_profit / total_revenue:.1%}")
```
